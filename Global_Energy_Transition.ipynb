```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("owid_energy.csv")

# Filter data from 2000 to 2025
energy_df = df[
    (df["year"] >= 2000) &
    (df["year"] <= 2025)
].copy()

print("Dataset loaded successfully!")
print("Shape:", energy_df.shape)
```

```python
# Calculate yearly average energy shares

yearly_trends = energy_df.groupby("year")[
    [
        "renewables_energy_share_pct",
        "fossil_fuels_energy_share_pct"
    ]
].mean().reset_index()

# Create professional graph
plt.figure(figsize=(14, 7))

plt.plot(
    yearly_trends["year"],
    yearly_trends["renewables_energy_share_pct"],
    color="#16A085",
    marker="o",
    linewidth=3,
    markersize=5,
    label="Renewable Energy"
)

plt.plot(
    yearly_trends["year"],
    yearly_trends["fossil_fuels_energy_share_pct"],
    color="#E74C3C",
    marker="s",
    linewidth=3,
    markersize=5,
    label="Fossil Fuels"
)

plt.title(
    "Global Energy Transition: Renewable vs Fossil Fuels",
    fontsize=17,
    fontweight="bold",
    pad=20
)

plt.xlabel("Year", fontsize=12, fontweight="bold")
plt.ylabel("Average Energy Share (%)", fontsize=12, fontweight="bold")

plt.xticks(range(2000, 2026, 2), rotation=45)
plt.legend(fontsize=11, loc="best")
plt.grid(alpha=0.25, linestyle="--")

plt.tight_layout()

# Save high-resolution graph
plt.savefig(
    "Graph_01_Renewable_vs_Fossil.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Graph 1: Renewable Energy vs Fossil Fuels (2000–2025)

This line chart compares the average renewable energy and fossil fuel shares between 2000 and 2025.

**Key Observations:**
- Fossil fuels maintained a much higher average share throughout the period.
- Renewable energy showed a gradual upward trend.
- Fossil fuel share showed a gradual downward trend.

**Conclusion:**
The chart indicates a gradual shift towards renewable energy, although fossil fuels remained dominant.

**Limitation:** Values represent unweighted averages of available country and regional observations, not global weighted energy shares.

```python
# Calculate yearly average solar energy share

solar_trend = energy_df.groupby("year")[
    "solar_energy_share_pct"
].mean().reset_index()

# Create solar energy trend graph

plt.figure(figsize=(13, 6))

plt.plot(
    solar_trend["year"],
    solar_trend["solar_energy_share_pct"],
    color="#F39C12",
    marker="o",
    linewidth=3,
    markersize=5,
    label="Solar Energy"
)

plt.title(
    "Solar Energy Trend (2000–2025)",
    fontsize=17,
    fontweight="bold",
    pad=20
)

plt.xlabel("Year", fontsize=12, fontweight="bold")
plt.ylabel("Average Solar Energy Share (%)", fontsize=12)

plt.xticks(range(2000, 2026, 2), rotation=45)
plt.legend()
plt.grid(alpha=0.25, linestyle="--")
plt.tight_layout()

plt.savefig(
    "Graph_02_Solar_Energy.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Graph 2: Solar Energy Trend (2000–2025)

This line chart illustrates changes in the average solar energy share between 2000 and 2025.

**Key Observations:**
- Solar energy share remained very low during the early years.
- Gradual growth became visible after 2010.
- Solar energy showed a stronger increase after 2020.
- The highest average share appears towards 2025.

**Conclusion:**
The trend indicates increasing adoption of solar energy over the study period.

**Limitation:** The values represent unweighted averages of available country and regional observations, not global weighted energy shares.
