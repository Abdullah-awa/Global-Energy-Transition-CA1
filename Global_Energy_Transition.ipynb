```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("owid_energy.csv")

# Filter data from 2000 to 2025
energy_df = df[
    (df["year"] >= 2000) &
    (df["year"] <= 2025)
].copy()

print("Dataset loaded successfully!")
print("Shape:", energy_df.shape)
```

```python
# Calculate yearly average energy shares

yearly_trends = energy_df.groupby("year")[
    [
        "renewables_energy_share_pct",
        "fossil_fuels_energy_share_pct"
    ]
].mean().reset_index()

# Create professional graph
plt.figure(figsize=(14, 7))

plt.plot(
    yearly_trends["year"],
    yearly_trends["renewables_energy_share_pct"],
    color="#16A085",
    marker="o",
    linewidth=3,
    markersize=5,
    label="Renewable Energy"
)

plt.plot(
    yearly_trends["year"],
    yearly_trends["fossil_fuels_energy_share_pct"],
    color="#E74C3C",
    marker="s",
    linewidth=3,
    markersize=5,
    label="Fossil Fuels"
)

plt.title(
    "Global Energy Transition: Renewable vs Fossil Fuels",
    fontsize=17,
    fontweight="bold",
    pad=20
)

plt.xlabel("Year", fontsize=12, fontweight="bold")
plt.ylabel("Average Energy Share (%)", fontsize=12, fontweight="bold")

plt.xticks(range(2000, 2026, 2), rotation=45)
plt.legend(fontsize=11, loc="best")
plt.grid(alpha=0.25, linestyle="--")

plt.tight_layout()

# Save high-resolution graph
plt.savefig(
    "Graph_01_Renewable_vs_Fossil.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Graph 1: Renewable Energy vs Fossil Fuels (2000–2025)

This line chart compares the average renewable energy and fossil fuel shares between 2000 and 2025.

**Key Observations:**
- Fossil fuels maintained a much higher average share throughout the period.
- Renewable energy showed a gradual upward trend.
- Fossil fuel share showed a gradual downward trend.

**Conclusion:**
The chart indicates a gradual shift towards renewable energy, although fossil fuels remained dominant.

**Limitation:** Values represent unweighted averages of available country and regional observations, not global weighted energy shares.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("owid_energy.csv")

# Filter data from 2000 to 2025
energy_df = df[
    (df["year"] >= 2000) &
    (df["year"] <= 2025)
].copy()

print("Dataset loaded successfully!")
print("Shape:", energy_df.shape)
```

```python
# Calculate average energy share for five indicators

energy_comparison = energy_df[
    [
        "renewables_energy_share_pct",
        "fossil_fuels_energy_share_pct",
        "solar_energy_share_pct",
        "wind_energy_share_pct",
        "hydro_energy_share_pct"
    ]
].mean()

# Create comparison bar chart

plt.figure(figsize=(12, 7))

colors = ["#27AE60", "#E74C3C", "#F1C40F", "#3498DB", "#9B59B6"]

bars = plt.bar(
    ["Renewables", "Fossil Fuels", "Solar", "Wind", "Hydro"],
    energy_comparison.values,
    color=colors,
    edgecolor="black",
    linewidth=0.7
)

plt.title(
    "Comparison of Energy Sources (2000–2025)",
    fontsize=17,
    fontweight="bold",
    pad=20
)

plt.xlabel("Energy Sources", fontsize=12, fontweight="bold")
plt.ylabel("Average Energy Share (%)", fontsize=12)

plt.bar_label(bars, fmt="%.2f%%", padding=4, fontsize=10)
plt.grid(axis="y", alpha=0.25, linestyle="--")
plt.tight_layout()

plt.savefig("Graph_04_Energy_Comparison.png",
            dpi=300, bbox_inches="tight")

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("owid_energy.csv")

# Filter data from 2000 to 2025
energy_df = df[
    (df["year"] >= 2000) &
    (df["year"] <= 2025)
].copy()

print("Dataset loaded successfully!")
print("Shape:", energy_df.shape)
```

```python
# Calculate yearly average energy shares

area_data = energy_df.groupby("year")[
    [
        "solar_energy_share_pct",
        "wind_energy_share_pct",
        "hydro_energy_share_pct",
        "other_renewables_including_bioenergy_energy_share_pct"
    ]
].mean().reset_index()

# Create stacked area chart

plt.figure(figsize=(14, 7))

plt.stackplot(
    area_data["year"],
    area_data["solar_energy_share_pct"],
    area_data["wind_energy_share_pct"],
    area_data["hydro_energy_share_pct"],
    area_data["other_renewables_including_bioenergy_energy_share_pct"],
    labels=["Solar", "Wind", "Hydro", "Other Renewables"],
    colors=["#F1C40F", "#3498DB", "#2ECC71", "#9B59B6"],
    alpha=0.85
)

plt.title(
    "Renewable Energy Mix Over Time (2000–2025)",
    fontsize=17,
    fontweight="bold",
    pad=20
)

plt.xlabel("Year", fontsize=12, fontweight="bold")
plt.ylabel("Average Energy Share (%)", fontsize=12)

plt.xticks(range(2000, 2026, 2), rotation=45)
plt.legend(loc="upper left")
plt.grid(alpha=0.2, linestyle="--")
plt.tight_layout()

plt.savefig(
    "Graph_05_Energy_Mix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Select energy indicators for correlation analysis

correlation_columns = [
    "renewables_energy_share_pct",
    "fossil_fuels_energy_share_pct",
    "solar_energy_share_pct",
    "wind_energy_share_pct",
    "hydro_energy_share_pct"
]

# Calculate correlation matrix
correlation_matrix = energy_df[correlation_columns].corr()

# Create professional heatmap
plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="RdYlBu_r",
    fmt=".2f",
    linewidths=0.8,
    square=True,
    vmin=-1,
    vmax=1,
    cbar_kws={"label": "Correlation Strength"}
)

plt.title(
    "Correlation Between Energy Indicators (2000–2025)",
    fontsize=16,
    fontweight="bold",
    pad=20
)

plt.xticks(rotation=35, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()

plt.savefig(
    "Graph_06_Energy_Correlation_Heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Graph 6: Correlation Between Energy Indicators (2000–2025)

This heatmap illustrates the correlation between five selected energy indicators using correlation coefficients.

**Key Observations:**
- Renewable energy and fossil fuels show a negative correlation of -0.69.
- Renewable energy and hydro show a strong positive correlation of 0.73.
- Solar and wind energy show a moderate positive correlation of 0.46.
- Fossil fuels and hydro show a negative correlation of -0.53.

**Conclusion:**
The results show different relationships between renewable and fossil fuel indicators. Renewable energy has a positive relationship with hydro, while fossil fuels show negative relationships with several renewable indicators.

**Limitation:** Correlation does not prove causation. The calculations use available observations and may be affected by missing data and regional aggregates.
