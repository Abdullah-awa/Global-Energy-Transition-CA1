```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# Visualisation style
sns.set_theme(style="whitegrid", palette="Set2")

print("All libraries imported successfully!")
```

```python
from pathlib import Path

file_path = Path.home() / "Downloads" / "owid_energy.csv"

print("File exists:", file_path.exists())
print("File location:", file_path)
```

```python
# Load the energy dataset

df = pd.read_csv("owid_energy.csv")

# Display first five rows
df.head()
```

```python
# Check dataset dimensions
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check dataset information
df.info()
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

# Display columns with missing values
missing_values = missing_values[missing_values > 0]

print("Total Missing Values:", df.isnull().sum().sum())

print("\nTop 15 Columns with Missing Values:")
print(missing_values.sort_values(ascending=False).head(15))
```

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
# Check the year range

print("Earliest Year:", df["year"].min())
print("Latest Year:", df["year"].max())

# Count unique years
print("Total Unique Years:", df["year"].nunique())

# Count unique countries and regions
print("Total Countries/Regions:", df["country"].nunique())
```

```python
# Filter data from 2000 to 2025

energy_df = df[
    (df["year"] >= 2000) &
    (df["year"] <= 2025)
].copy()

# Check the filtered dataset

print("Filtered Dataset Shape:", energy_df.shape)
print("Start Year:", energy_df["year"].min())
print("End Year:", energy_df["year"].max())

energy_df.head()
```

```python
# Identify important energy indicators

energy_columns = [
    "renewables_energy_share_pct",
    "fossil_fuels_energy_share_pct",
    "solar_energy_share_pct",
    "wind_energy_share_pct",
    "hydro_energy_share_pct"
]

# Check which columns are available
for column in energy_columns:
    if column in energy_df.columns:
        print("Available:", column)
    else:
        print("Not found:", column)
```

```python
# Analyse missing values in selected energy indicators

selected_columns = [
    "renewables_energy_share_pct",
    "fossil_fuels_energy_share_pct",
    "solar_energy_share_pct",
    "wind_energy_share_pct",
    "hydro_energy_share_pct"
]

missing_report = pd.DataFrame({
    "Missing Values": energy_df[selected_columns].isnull().sum(),
    "Missing Percentage": (
        energy_df[selected_columns].isnull().mean() * 100
    ).round(2)
})

missing_report
```

```python
# Statistical summary of selected energy indicators

energy_df[selected_columns].describe().round(2)
```

```python
# Check yearly data availability

yearly_data = energy_df.groupby("year")[selected_columns].count()

yearly_data.head(10)
```
